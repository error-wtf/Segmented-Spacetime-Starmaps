## 🎯 Usage Guide

### Features:

**1. Sky Maps**
- 🗺️ 2D Sky Map: Full sky projection with 500k stars
- 🌐 3D Sky Map: Interactive 3D view (rotate with mouse)
- 🔍 Constellation View: Focus on specific regions

**2. SSZ Physics**
- 📊 g₁/g₂ Domains: Segment density structure
- ⏱️ Time Dilation: SSZ vs GR comparison
- 🌀 Radial Stretch: Metric deviation

**3. Object Search**
- Search by name: `Sag A*`, `Betelgeuse`, `M31`
- Search by coordinates: `266.4, -29.0`
- View detailed physics for any object

**4. Data Fetch**
- Enrich database with AKARI infrared data
- ESO/ALMA spectroscopy integration
- Real-time enrichment statistics

## 🐛 Troubleshooting

**If Gradio link doesn't appear:**
1. Runtime → Restart runtime
2. Run all cells again
3. Wait 2-3 minutes

**If app crashes:**
- Colab free tier has limited RAM
- Try closing other Colab notebooks
- Restart runtime and try again

**If plots don't load:**
- Check browser console for errors
- Try refreshing the Gradio page
- Some plots require object selection first
