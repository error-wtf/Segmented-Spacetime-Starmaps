# 📖 User Guide

## 🎮 Using the App

Once the Gradio link appears above, click it to open the interactive dashboard.

### **Tab 1: Object Search & Selection**

1. **Search Box:** Type object name, GAIA ID, or coordinates
   - Examples: `"Betelgeuse"`, `"M87"`, `"Sgr A*"`, `"4057468631175006336"`
   - Coordinates: `"12.5, +45.3"` (RA, Dec in degrees)

2. **Dropdown:** Select from results

3. **Auto-Update:** All plots update automatically when you select an object

### **Tab 2: Visualizations**

#### **2D Sky Map (First Sub-Tab)**
- Shows all 228,661 stars in RA/Dec coordinates
- **Click** on any star to select it
- **Zoom:** Scroll wheel
- **Pan:** Click and drag
- Yellow star with red border = selected object

#### **3D Sky Map (Second Sub-Tab)**
- Galactic 3D view with Cartesian coordinates
- **Sliders:** Adjust camera distance, horizontal angle, vertical angle
- **"Center on Selected Object"** button: Zoom into selected star
- **"Reset View"** button: Return to default galactic view
- High Quality checkbox: More stars (slower on Colab)

#### **Constellation View (Third Sub-Tab)**
- Focused view on specific sky regions
- **FOV Slider:** Field of view (5° - 45°)
- **RA/Dec Inputs:** Center coordinates
- Shows objects within FOV circle

### **Tab 3: Physics Plots**

All plots use the **currently selected object** from Tab 1.

#### **Domain Structure**
- Shows SSZ segmentation zones:
  - **g₃** (violet): Core collapse zone (r < r_c)
  - **g₂** (orange): Transition zone (r_c < r < 2r_c)
  - **g₁** (blue): Stable zone (r > 2r_c)
- Toggle "Show Objects" to see famous objects on the plot

#### **Time Dilation**
- Compares **SSZ** (blue) vs **GR** (red dashed)
- Formula: A(r) = D(r)·(1 - r_s/r) for SSZ
- Shows where SSZ = GR (yellow marker)

#### **Radial Stretch**
- Shows B(r) spatial metric component
- SSZ (blue) vs GR (red dashed)

#### **Combined Analysis**
- 4-panel overview:
  1. Time Dilation A(r)
  2. Radial Stretch B(r)
  3. Segmentation γ(r)
  4. Collapse Rate C(r) = -dγ/dr

#### **SEG Performance**
- Shows SSZ vs GR near Schwarzschild radius
- Only meaningful for compact objects (black holes, neutron stars)
- Normal stars are too far from r_s (plot shows warning)

### **Tab 4: Object Data**

- Shows full GAIA DR3 data for selected object
- Includes: coordinates, proper motion, parallax, magnitudes, temperatures
- Copy-paste friendly format

## 📚 Scientific Background

### **SSZ Theory**

Segmented Spacetime (SSZ) proposes that spacetime is **segmented** around massive objects, with:

- **φ-driven dynamics** (φ = golden ratio ≈ 1.618)
- **Three distinct zones** (g₃, g₂, g₁)
- **Sharp transitions** at r_c = φ·r_s
- **Continuous metric** (C² differentiable)

### **Key Differences from General Relativity**

| Property | GR | SSZ |
|----------|----|----|
| **Singularity** | r = 0 | Resolved via segmentation |
| **Event Horizon** | r = r_s | Modified A(r) at r_s |
| **Photon Sphere** | r = 1.5r_s | Shifted due to D(r) |
| **ISCO** | r = 6r_s (Schwarzschild) | Modified by SSZ domains |
| **Time Dilation** | A = 1 - r_s/r | A = D(r)·(1 - r_s/r) |

### **Observable Predictions**

- **Black hole shadows:** Slightly different from GR
- **Gravitational waves:** Modified ringdown frequencies
- **Perihelion precession:** Matches GR in weak field
- **Light bending:** Domain-dependent corrections
