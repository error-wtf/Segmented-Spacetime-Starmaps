## 🎯 Features in the App:

### **📊 Time Dilation Plot:**
- Shows **A(r) = D(r)·(1-r_s/r)** for SSZ
- Compares with **GR: A(r) = 1-r_s/r**
- **Yellow marker**: Intersection point SSZ ∩ GR
- **Sharp break** at r_c with domain shading

### **📈 Segmentation Field γ(r):**
- Shows **γ(r) = 1 - α·exp(-(r/r_c)²)**
- **Collapse Rate** C(r) = -dγ/dr in second panel
- **Piecewise fits** for g₂ (collapse) and g₁ (stable)

### **🌐 Combined Analysis:**
- 4-panel comparison
- All domains visualized
